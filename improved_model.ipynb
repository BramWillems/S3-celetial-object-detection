```python
from ultralytics import YOLO

model = YOLO("yolov8m.pt")

results = model.train(
    data="Data/DeepSpaceYoloDataset/data.yaml",
    epochs=200,     # Previously 100, but increased to 200 for better performance
    imgsz=608,      # This is the image size based on the dataset
    batch=8,        # Random batch size, bigger uses more GPU memory, this seems to work
    patience=20,    # Early stopping after 20 epochs without improvement, this is to prevent overfitting and save time

    # Data augmentation settings
    degrees=180.0,  # Randomly rotate images by up to 180 degrees, which is important for space object detection as they can be in any orientation due to the earth spinning around its axis
    translate=0.1,  # Randomly translate images by up to 10% (Move right left up or down), this makes it so the model does not learn patterns based on position
    scale=0.5,      # Randomly zoom images, makes it so that the model can detects smaller objects easier
    fliplr=0.5,     # 50% chance to flip an image left or right, this is for the same reason as rotation, space objects can be in any orientation and this makes it so the model does not learn patterns based on orientation
    flipud=0.5,     # 50% chance to flip an image up or down, this is for the same reason as rotation, space objects can be in any orientation and this makes it so the model does not learn patterns based on orientation

    # Colour settings
    hsv_h=0.01,     # Small change in hue, this is because the classes can vary in colour but the colour is still important for detection
    hsv_s=0.0,      # I decided to not change the saturation, because i colours are important but I do not fully understand how this affects the images
    hsv_v=0.4,      # Randomly changes the birghtness of the images, this is important because the model had difficulties with detecting suble differences in brightness

    # Other augmentations
    mosaic=1.0,     # Mosaic augmentation increases recall, further explained in report.
    mixup=0.1       # adds a 10% chance to mix two images together, this is for the same reason as mosaic, it increases recall
)
```
